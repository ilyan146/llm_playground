```python
from huggingface_hub import login
from transformers import pipeline
import os
```

```python
import torch
```

```python
print(torch.cuda.is_available())
```

```python
# Login for auth
login(token=os.getenv("HF_API_KEY"))
```

```python
# ner = pipeline("ner", device="cuda")

ner = pipeline("ner", model="autoevaluate/entity-extraction", device="cuda")
result = ner("Hugging face is a great platform for AI Engineers such as John to get acces to OSS models.")
for entity in result:
  print(entity)
```

```python
classifier = pipeline(task="zero-shot-classification", model="MoritzLaurer/deberta-v3-large-zeroshot-v2.0", device="cuda") # No examples given
result = classifier("Hugging Face pipelines are a high level API for inference of LLMs with common tasks.", candidate_labels=["education", "sports", "technology"])
print(result)
```
